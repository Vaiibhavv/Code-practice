```python
## find the second highest number 


def second_highest(arr):
    first=second=None

    for num in arr:
        if num==first or num==second:
            continue
        if first is None or num>first:
            second=first
            first=num
        elif second is None or num>second:
            second=num
    return second if second is not None else first
       
arr=[1, 3, 5, 2,7, 4]
print(second_highest(arr))
```

```python
## check the number is prime or not 
import math
def prime_number(num):
    sq=math.sqrt(num)
    if num==1:
        return "number is prime"
    else:
        is_prime=True
        for i in range(2,int(sq)+1):
            if num%i==0:
                is_prime=False
                break
        if is_prime:
            print("number is prime__")
        else:
            print("number is not prime")

prime_number(12)
        

def prime_2(n2):
    cnt=0
    for i in  range(1, n2+1):
        if n2%i==0:
            cnt+=1
    if cnt==2:
        print("number is prime")
    else:
        print("number is not prime")
prime_2(12)

    
```

```python
# print the diamond pattern
def diamond_pattern(n):
    for i in range(1, n):
        space=n-i
        star=2*i-1
        print(" "*space +"*" *star)

    for j in range(n,0,-1):
        space=n-j
        star=2*j-1
        print(" "*space + "*" * star)
diamond_pattern(4)
```

```python
## hollow pattern 
def hollow(n):
    for i in range(n):
        for j in range(n):
            if i==0 or i==n-1 or j==0 or j==n-1:
                print("*", end="")
            else:
                print(" ",end="")
        print()
hollow(5)
```

```python
## sum should be single number 
def single_digit(n):
    while n>10:
        total=0
        while n>0:
            total+=n%10
            n=n//10
        n=total
    return total
print(single_digit(67))
```

```python
## two sum 
# the first approach is - List= [2,7,11,15], target=9, answer [0,1]
# the first approach is - we can solve this problem using the O(n square solution) , where we will find 
# all posible solutions 

def two_sum_1(n1, target):
    sum=0
    for i in range(len(n1)):
        first=n1[i]
        for j in range(i+1, len(n1)):
            second=n1[j]
            sum=first+second
            if sum==target:
                return [i,j]
n1=[2,7,11,15]
target=9
print(two_sum_1(n1,target))

"""second approach is two pointer approach is the array is sorted , left=minimum means the 0th position 
of an array and right is the maximum number """ 

def two_sum_2(n1, target):
    ## first sort the array 
    n1.sort()
    left_p=0
    right_p=len(n1)-1

    #check until the left is less than right pointer 
    while (left_p<right_p):
        sum=n1[left_p]+n1[right_p]

        # now check whether the sum is less than or greter than the current calculated sum , if sum
        """ if sum is less than increase the left pointer else decrease the right pointer """
        if sum<target:
            left_p+=1
        elif sum>target:
            right_p-=1
        else:
            return [left_p,right_p]

print(two_sum_2(n1,target))


## next we can solve using the hashing alogrithm, it will give us t O(n ) time complexity 
def two_sum_3(n1,target):
    #  first+second=target=  we can say that second=target-first 
    seen_dict={}
    for i in range(len(n1)):
        first=n1[i]
        second=target-first

        if second in seen_dict:
            return [i,seen_dict[second]] ## value of the second element means the index 
        seen_dict[first]=i
print(two_sum_3(n1,target))


```

```python
## rotate array , 
# , here will rotate an array , for range in kth time
# input- [1,2,3,4,5,6,7] , output= [5, 6, 7, 1, 2, 3, 4]

def rotate_array_1(re_array,k_):
    for i in range(k_):
        last_element=re_array.pop()
        re_array.insert(0,last_element)
    return re_array
re_array=[1,2,3,4,5,6,7]
k_=3 
print(rotate_array_1(re_array,k_))

## the second approach is using the reverse the array , and then again reverse it, and again reverse
def rotate_array_main(re_array,k_):
    lengt=len(re_array)
    k_=k_%lengt
    def rotate_array_2(first, last):
        while (first<last):
            re_array[first],re_array[last]=re_array[last],re_array[first]
            first+=1
            last-=1
    rotate_array_2(0,lengt-1)
    rotate_array_2(0,k_-1)
    rotate_array_2(k_,lengt-1)
    return re_array
lis=[1,2,3,4,5,6,7]
kk=3
print(rotate_array_main(lis,kk))


        
```

```python
## intersection of two array, find the common element in both the arrays 

#Input: nums1 = [1,2,2,1], nums2 = [2,2]
#Output: [2] 

## first approahc is using the hashmap
def intersect_twoArray(a1, a2):
    seen=set(a1)
    dic=set()
    for i in a2:
        if i in seen:
            dic.add(i)
    return list(dic) 
a1=[1,2,2,1]
a2=[2,2]
print(intersect_twoArray(a1,a2))
```
