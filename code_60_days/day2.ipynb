```python
employees = [
    {"id":1,"dept":"IT","salary":50000},
    {"id":2,"dept":"HR","salary":30000},
    {"id":3,"dept":"IT","salary":70000},
    {"id":4,"dept":"HR","salary":60000},
    {"id":5,"dept":"Finance","salary":90000},
]

# expected output 

# {
#     "IT":{
#         "count":2,
#         "total_salary":120000,
#         "average_salary":60000
#     }
# }

result={}
for emp in employees:
    dept=emp['dept']
    salary=emp['salary']

    if dept not in result:
        result[dept]={"count":0,"total_salary":0}

    result[dept]['count']+=1
    result[dept]['total_salary']+=salary
print(result)

## compute the average 
for dept in result:
    result[dept]['average_salary']=(result[dept]['total_salary']//result[dept]['count'])
print(result)
```

```python
## remove the duplicate from records 

records = [
    {"id":1,"name":"A"},
    {"id":2,"name":"B"},
    {"id":1,"name":"A"},
    {"id":3,"name":"C"},
]

resultlist=[]
rs={}
for items in records:
    id=items['id']
    if id not in rs:
        rs[id]={"count":0}
    rs[id]['count']+=1
    if rs[id]['count']==1:
        resultlist.append(items)
print(resultlist)

    

```

```python
## lambda function 
numbers = [5,2,8,1]

sorted(numbers, key=lambda x: x)

```

### lambda function with map, filter and reduce

```python
#mulitply 2 with every item in iterable object (numbers)

result=list(map(lambda x:x*2,numbers))
print(result)
```

```python
# filter (find out the odd number from list )
# In filter the function will iterate on each item of the list and return the items which are true
numbers1 = [1, 2, 3, 4, 5, 6, 7, 8]
odd_result= list(filter(lambda x:x%2!=0, numbers1))
print(odd_result)
```

```python
# reduce, 
"""(it generally used when we want to generate a single result from iteratble object eg, from a 
list geneate the total sum of the list ( like cummalative sum)"""

from functools import reduce

total_sum=reduce(lambda x,y:x+y,numbers1)
print(total_sum)
multiply= reduce(lambda x,y:x*y,numbers)
print(multiply)
```
