```python
"""Find largest / smallest element in array"""
"""   
approach 1 ( by using built in functions) """
array1=[1, 2, 3, 4, 81, 9]
#print("maximum number", max(array1))
#print("minimum number",min(array1))

# approch2 by using iterative or logical approach

largest=array1[0]
smallest=array1[0]

for num in array1[1:]:
    if num>largest:
        largest=num
    if num<smallest:
        smallest=num
print("largest number", largest)
print("smalles number", smallest)

print("third largest element in an array")

def thirdmax(nums):
    # assign the first , second and third variable
    first=second=third=None

    # iterate over the loop
    for num in nums:
        ## checking the duplicate
        if num==first or num==second or num==third:
            continue

        ## actual iteration for calculating the first, second and third number
        if first is None or num>first:
            third=second
            second=first
            first=num
        elif second is None or num>second:
            third=second
            second=num
        elif third is None or num>third:
            third=num
        
    return third if third is not None else first
nums=array1=[1, 2, 3, 4, 81, 9]
print(thirdmax(nums))
```

```python
# Check if two strings are anagrams  ( anagram is all about checking the frequency not the order)
string1="race" 
string2="care"

#print(sorted(string1)) ## by default sorted print the output as a list 
"""approach 1 by using sorted function and join the list at the end"""
#s1= sorted(string1)
#s2= sorted(string2)
a1="".join(sorted(string1))
a2="".join(sorted(string2))
if len(a1)==len(a2) and a1==a2:
    print("both strings are anagram")
else:
    print("both strings are not anagram")


"""approach 2 by using the Counter"""

from collections import Counter
def check_anangram(s1,s2):
    if len(s1)!=len(s2):
        return False
    elif Counter(s1)==Counter(s2):
        return "Strings are anagram"
    
check_anangram("race", "care")

"""Approach 3 by using manual counter"""

def check_anangram_manual(s1, s2):
    if len(s1)!=len(s2):
        return False
    
    f={}
    for x in s1:
        f[x]=f.get(x, 0)+1

    for x in s2:
        if x not in f:
            return False
        
        f[x]-=1
        if f[x]<0:
            return False
    return True
print(check_anangram_manual("race","care"))
    
```

```python
## Vowels Count and  → LC 345 – Reverse Vowels

"""Approach 1 (common apporach) """
aWord="vaibhav"
vowels_array=['a','e','i','o','u']
count=0
for char in aWord:
    if char in vowels_array:
        count+=1
    else:
        pass

print("Vowels count in string",count)

"""Reverse ONLY the vowels, keep everything else in the same place.
eg- hello ---> holle"""

## approach 1 (two pointers method)

def reverseVowels(wordA):
    # first convert the string to list , because string is immutable
    lword=list(wordA)
    vowels_l= set('aeiouAeiou')
    left=0
    right=len(lword)-1

    #iterate on lword
    while left < right:
        while left<right and lword[left] not in vowels_l:
            left+=1
        while left<right and lword[right] not in vowels_l:
            right-=1

        ## in outer loop above condition failed so we got the vowels, now will swap them
        lword[left],lword[right]=lword[right],lword[left]

        left+=1
        right-=1 
    return "".join(lword)
print(reverseVowels("hello"))
```

```python
#Hollow Square
"""
*****
*   *
*   *
*   *
*****

"""

def hollow_square(n):
    for i in range(n):
        for j in range(n): ## for each iteration of i, the j will run all ites (5 ) iteration
            if i==0 or i==n-1 or j==0 or j==n-1:
                print("*",end="")
            else:
                print(" ",end="")
        print() 
hollow_square(5)
```

```python
#19.	Sum of Digits → LC 258 – Add Digits
"""in this example you have to print the sum of numbers, until the single digit
eg. sum(156)=12 , sum(12)=3"""

#-first approach (straight forward) (complexity - number of iteration) complexit  (O(d × iterations)) and space O(11)
def numSum(num):
    while num>=10:
        total=0
        while num>0:
            total+=num%10
            num=num//10
        num=total
    return total
print(numSum(156))

## second approach (digital root)
""" 1+(num-1)%9 , return single digit sum, conditon num==0 and also for num%9==0 then answer will be 9"""

def numsumDigitalRoot(num):
    if num==0:
        return 0 
    return 1+(num-1)%9
print(numsumDigitalRoot(54))

"""approach 3 """

def sumdigit(num):
    while num>=10:
        ## sum is work only on iterator
        num=sum(int(i) for i in str(num))
    return num 
print(sumdigit(156))
```

```python
# Two sum 
"""Input: nums = [2,7,11,15], target = 9
Output: [0,1]
Explanation: Because nums[0] + nums[1] == 9, we return [0, 1]"""

from typing import List

def twoSumSolution(nums:List[int],target:int):
    for i in range(len(nums)):
        for j in range(i+1,len(nums)):
            if nums[i]+nums[j]==target:
                return [i,j]
            else:
                None
inputL=[2,7,11,15]
target=9
print(twoSumSolution(inputL,target))

```

```python
# two sum = for sorted array
"""if array is sorted, then two pointer approach will be the better option 
  O(n) and  space O(1)  
  List= [2,7,11,15], target=9, answer [0,1]
  """

def twoSumSortedArray(nums,target):
    left=0
    right=len(nums)-1
    while left<right:
        current_sum=nums[left]+nums[right]
        
        if current_sum==target:
            return [left,right]
        elif current_sum<target:
            left+=1 
        else:
            right-=1
        
inputL=[2,7,11,15]
target=9
print(twoSumSortedArray(inputL,target))


## second approach by hash map() dict  ( hash map use)- 
"""complement ( to find a number by using a number eg. target= 100, current=40 , 
so complement=target-current, is 60)
frequency
seen before
duplicate detection
prefix sum
"""
def hashmapTwosum(arr,target):
    seen_dict={}

    for i,num in enumerate(arr):
        required_target=target-num
        
        if required_target in seen_dict:
            return [seen_dict[required_target],i]
        seen_dict[num]=i 
    return seen_dict
inputL=[2,7,11,15]
target=9
print(hashmapTwosum(inputL,target))  



```

```python
#Moves Zero
"""Given an integer array nums, move all 0's to the end of it while maintaining 
   the relative order of the non-zero elements.

Note that you must do this in-place without making a copy of the array.
eg  input [1,0,3,0,2,] , output = [1,3,2,0,0,0]
 """
def movesZero(arr):
    zero=0
    zerolist=[]
    for i in arr:
        if i==zero:
            zerolist.append(i)
            arr.remove(i)
    return arr+zerolist
input_arr=[1,0,3,0,2]
print(movesZero(input_arr)) , ## this is slower (0(n)square)

##  By using two pointer approach, 
"""iterate through the array and maintain a pointer indicating where the next non-zero 
element should go. Whenever I encounter a non-zero value, I swap it with that position 
and move the pointer forward. This keeps non-zero elements 
in order while pushing zeros to the end."""

def moveZeroTwoPointer(arr):
    swap_position=0

    for i in range(len(arr)):
        if arr[i]!=0:
            arr[swap_position],arr[i]=arr[i],arr[swap_position]
            swap_position+=1
    return arr 

new_input_arr=[1,0,3,0,2]
print(moveZeroTwoPointer(new_input_arr))
```

```python
## Rotate Array 
""" 
Given an integer array nums, rotate the array to the right by k steps, where k is non-negative.
Example 1:
Input: nums = [1,2,3,4,5,6,7], k = 3
Output: [5,6,7,1,2,3,4]
Explanation:
rotate 1 steps to the right: [7,1,2,3,4,5,6]
rotate 2 steps to the right: [6,7,1,2,3,4,5]
rotate 3 steps to the right: [5,6,7,1,2,3,4]

Example 2:

Input: nums = [-1,-100,3,99], k = 2
Output: [3,99,-1,-100]
Explanation: 
rotate 1 steps to the right: [99,-1,-100,3]
rotate 2 steps to the right: [3,99,-1,-100], come up with best approach and 
"""

# Pattern Recognition

# This problem teaches array reversal tricks, which appear in:

# Rotate array

# Reverse words in string

# Palindrome checks

# Matrix rotation

## solution
# - 1st straight forward approach (time complexity is O(n))
def rotateArray(nums, k):
    n = len(nums)
    k = k % n
    return nums[-k:] + nums[:-k]
array=[1,2,3,4,5,6,7]
k=3
print(rotateArray(array,k=k))

## 2 second approach (optimal solution time complexity O(n)) and space complexity(O(1))

def rotate(nums, k):
    l=len(nums)
    k=k%l

    def rotateReverse(first,last):
        while(first<last):
            nums[first],nums[last]=nums[last],nums[first]
            first+=1
            last-=1
    rotateReverse(0,l-1) ## full reverse the list
    rotateReverse(0,k-1) ## only first kth elements
    rotateReverse(k,l-1) ## reverse remaining ones after kth
    return nums

array_=[1,2,3,4,5,6,7]
k_=3
print(rotate(array_,k_))


#----------------------------------------------------------------
# 3rd Approach (rotate the array kth time) , the time complexity (O*k   )

def rotateKthtime(nums, k):
    n=len(nums)

    for _ in range(k): # will rotate kth time
        last_element=nums.pop()
        nums.insert(0,last_element)
    return nums
array_=[1,2,3,4,5,6,7]
k_=3
print(rotateKthtime(array_,k=k_))




```
