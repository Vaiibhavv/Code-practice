```python
"""Find largest / smallest element in array"""
"""   
approach 1 ( by using built in functions) """
array1=[1, 2, 3, 4, 81, 9]
#print("maximum number", max(array1))
#print("minimum number",min(array1))

# approch2 by using iterative or logical approach

largest=array1[0]
smallest=array1[0]

for num in array1[1:]:
    if num>largest:
        largest=num
    if num<smallest:
        smallest=num
print("largest number", largest)
print("smalles number", smallest)

print("third largest element in an array")

def thirdmax(nums):
    # assign the first , second and third variable
    first=second=third=None

    # iterate over the loop
    for num in nums:
        ## checking the duplicate
        if num==first or num==second or num==third:
            continue

        ## actual iteration for calculating the first, second and third number
        if first is None or num>first:
            third=second
            second=first
            first=num
        elif second is None or num>second:
            third=second
            second=num
        elif third is None or num>third:
            third=num
        
    return third if third is not None else first
nums=array1=[1, 2, 3, 4, 81, 9]
print(thirdmax(nums))
```

```python
## Sum of elements in an array

"approach 1 using built in functions"

array1=[1, 2, 3, 4, 81, 9]
sum_of= sum(array1)
print("sum of all element in an array", sum_of)


"""approach 2"""
sum=0
for i in array1:
    sum=sum+i
print("total sum is ", sum)

# running sum  in normal form
for i in array1:
    sum=sum+i
    print(sum, "\n")
    

"""approach 3- running sum in list format ( complexity is O(N) both space and time) """ 

def runningSum(arr1):
    result=[]
    total=0
    for i in arr1:
        total+=i
        result.append(total)
    return result
arr_list=[1, 2, 3, 4, 5]
print(runningSum(arr_list))

"""approach 3 (better approac space complexity is O(1))"""
def runningSumOpt(arr1):
    for i in range(1, len(arr1)):
        arr1[i]=arr1[i]+arr1[i-1]
    return arr1

arr_list=[1, 2, 3, 4, 5]
print(runningSumOpt(arr_list))

```

```python
#Find duplicate elements in an array

"""approach 1"""
#list_1=[2, 4, 55, 7, 4, 5,2]
list_1=[1, 1, 1, 1]
f={}
duplicate_list=[]
for x in list_1:
    f[x]=f.get(x,0)+1
    #if f[x]>1: 
    ## or 
    if f[x]==2:
        duplicate_list.append(x)

print(f)
print(duplicate_list)
#print(set(duplicate_list)) ## because it will remove the duplicate


"""approach 2 (using set)"""

seen=set()
duplicate=set()
for x1 in list_1:
    if x1 in seen:
        seen.add(x1)
    else:
        duplicate.add(x1)
print(list(duplicate))
```

```python
# Check if two strings are anagrams  ( anagram is all about checking the frequency not the order)
string1="race" 
string2="care"

#print(sorted(string1)) ## by default sorted print the output as a list 
"""approach 1 by using sorted function and join the list at the end"""
#s1= sorted(string1)
#s2= sorted(string2)
a1="".join(sorted(string1))
a2="".join(sorted(string2))
if len(a1)==len(a2) and a1==a2:
    print("both strings are anagram")
else:
    print("both strings are not anagram")


"""approach 2 by using the Counter"""

from collections import Counter
def check_anangram(s1,s2):
    if len(s1)!=len(s2):
        return False
    elif Counter(s1)==Counter(s2):
        return "Strings are anagram"
    
check_anangram("race", "care")

"""Approach 3 by using manual counter"""

def check_anangram_manual(s1, s2):
    if len(s1)!=len(s2):
        return False
    
    f={}
    for x in s1:
        f[x]=f.get(x, 0)+1

    for x in s2:
        if x not in f:
            return False
        
        f[x]-=1
        if f[x]<0:
            return False
    return True
print(check_anangram_manual("race","care"))
    
```
