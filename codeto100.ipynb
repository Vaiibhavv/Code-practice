```python
## Vowels Count and  → LC 345 – Reverse Vowels

"""Approach 1 (common apporach) """
aWord="vaibhav"
vowels_array=['a','e','i','o','u']
count=0
for char in aWord:
    if char in vowels_array:
        count+=1
    else:
        pass

print("Vowels count in string",count)

"""Reverse ONLY the vowels, keep everything else in the same place.
eg- hello ---> holle"""

## approach 1 (two pointers method)

def reverseVowels(wordA):
    # first convert the string to list , because string is immutable
    lword=list(wordA)
    vowels_l= set('aeiouAeiou')
    left=0
    right=len(lword)-1

    #iterate on lword
    while left < right:
        while left<right and lword[left] not in vowels_l:
            left+=1
        while left<right and lword[right] not in vowels_l:
            right-=1

        ## in outer loop above condition failed so we got the vowels, now will swap them
        lword[left],lword[right]=lword[right],lword[left]

        left+=1
        right-=1 
    return "".join(lword)
print(reverseVowels("hello"))
```

```python
# Two sum 
"""Input: nums = [2,7,11,15], target = 9
Output: [0,1]
Explanation: Because nums[0] + nums[1] == 9, we return [0, 1]"""

from typing import List

def twoSumSolution(nums:List[int],target:int):
    for i in range(len(nums)):
        for j in range(i+1,len(nums)):
            if nums[i]+nums[j]==target:
                return [i,j]
            else:
                None
inputL=[2,7,11,15]
target=9
print(twoSumSolution(inputL,target))

```

```python
# two sum = for sorted array
"""if array is sorted, then two pointer approach will be the better option 
  O(n) and  space O(1)  
  List= [2,7,11,15], target=9, answer [0,1]
  """

def twoSumSortedArray(nums,target):
    left=0
    right=len(nums)-1
    while left<right:
        current_sum=nums[left]+nums[right]
        
        if current_sum==target:
            return [left,right]
        elif current_sum<target:
            left+=1 
        else:
            right-=1
        
inputL=[2,7,11,15]
target=9
print(twoSumSortedArray(inputL,target))


## second approach by hash map() dict  ( hash map use)- 
"""complement ( to find a number by using a number eg. target= 100, current=40 , 
so complement=target-current, is 60)
frequency
seen before
duplicate detection
prefix sum
"""
def hashmapTwosum(arr,target):
    seen_dict={}

    for i,num in enumerate(arr):
        required_target=target-num
        
        if required_target in seen_dict:
            return [seen_dict[required_target],i]
        seen_dict[num]=i 
    return seen_dict
inputL=[2,7,11,15]
target=9
print(hashmapTwosum(inputL,target))  



```

```python
#Moves Zero
"""Given an integer array nums, move all 0's to the end of it while maintaining 
   the relative order of the non-zero elements.

Note that you must do this in-place without making a copy of the array.
eg  input [1,0,3,0,2,] , output = [1,3,2,0,0,0]
 """
def movesZero(arr):
    zero=0
    zerolist=[]
    for i in arr:
        if i==zero:
            zerolist.append(i)
            arr.remove(i)
    return arr+zerolist
input_arr=[1,0,3,0,2]
print(movesZero(input_arr)) , ## this is slower (0(n)square)

##  By using two pointer approach, 
"""iterate through the array and maintain a pointer indicating where the next non-zero 
element should go. Whenever I encounter a non-zero value, I swap it with that position 
and move the pointer forward. This keeps non-zero elements 
in order while pushing zeros to the end."""

def moveZeroTwoPointer(arr):
    swap_position=0

    for i in range(len(arr)):
        if arr[i]!=0:
            arr[swap_position],arr[i]=arr[i],arr[swap_position]
            swap_position+=1
    return arr 

new_input_arr=[1,0,3,0,2]
print(moveZeroTwoPointer(new_input_arr))
```

```python
## Rotate Array 
""" 
Given an integer array nums, rotate the array to the right by k steps, where k is non-negative.
Example 1:
Input: nums = [1,2,3,4,5,6,7], k = 3
Output: [5,6,7,1,2,3,4]
Explanation:
rotate 1 steps to the right: [7,1,2,3,4,5,6]
rotate 2 steps to the right: [6,7,1,2,3,4,5]
rotate 3 steps to the right: [5,6,7,1,2,3,4]

Example 2:

Input: nums = [-1,-100,3,99], k = 2
Output: [3,99,-1,-100]
Explanation: 
rotate 1 steps to the right: [99,-1,-100,3]
rotate 2 steps to the right: [3,99,-1,-100], come up with best approach and 
"""

# Pattern Recognition

# This problem teaches array reversal tricks, which appear in:

# Rotate array

# Reverse words in string

# Palindrome checks

# Matrix rotation

## solution
# - 1st straight forward approach (time complexity is O(n))
def rotateArray(nums, k):
    n = len(nums)
    k = k % n
    return nums[-k:] + nums[:-k]
array=[1,2,3,4,5,6,7]
k=3
print(rotateArray(array,k=k))

## 2 second approach (optimal solution time complexity O(n)) and space complexity(O(1))

def rotate(nums, k):
    l=len(nums)
    k=k%l

    def rotateReverse(first,last):
        while(first<last):
            nums[first],nums[last]=nums[last],nums[first]
            first+=1
            last-=1
    rotateReverse(0,l-1) ## full reverse the list
    rotateReverse(0,k-1) ## only first kth elements
    rotateReverse(k,l-1) ## reverse remaining ones after kth
    return nums

array_=[1,2,3,4,5,6,7]
k_=3
print(rotate(array_,k_))


#----------------------------------------------------------------
# 3rd Approach (rotate the array kth time) , the time complexity (O*k   )

def rotateKthtime(nums, k):
    n=len(nums)

    for _ in range(k): # will rotate kth time
        last_element=nums.pop()
        nums.insert(0,last_element)
    return nums
array_=[1,2,3,4,5,6,7]
k_=3
print(rotateKthtime(array_,k=k_))




```

```python
# Merge sorted array, 
""" Problem statement is -merge  the two  sorted array in descending order. 
eg. - Input: nums1 = [1,2,3,0,0,0], m = 3, nums2 = [2,5,6], n = 3
Output: [1,2,2,3,5,6]
Explanation: The arrays we are merging are [1,2,3] and [2,5,6].
The result of the merge is [1,2,2,3,5,6] with the underlined elements coming from nums1.
and the final array should be stored in nums1
"""

def MergeSortedArray(nums1, nums2,m,n):
    #nums1=nums1[:m]+nums2[:n]
    ## use of extend method
    l=len(nums1)
    rm=l-m
    nums1.extend(nums2)
    nums1.sort()
    return nums1[m:]

nums1=[1,2,3,0,0,0]
nums2=[2,5,6]
m=3 
n=3 
print(MergeSortedArray(nums1,nums2,m,n))

## best approach O(m+n )  ( sort from the end with largest element)
def MergeSortedArraySo2l(nums1, nums2,m,n):
    p1 = m - 1  ## first array 
    p2 = n - 1  ## second array
    p = m + n - 1 ## length of first array= both combine array length

    while p1 >= 0 and p2 >= 0:

        if nums1[p1] > nums2[p2]:
            nums1[p] = nums1[p1]
            p1 -= 1
        else:
            nums1[p] = nums2[p2]
            p2 -= 1

        p -= 1

    while p2 >= 0:
        nums1[p] = nums2[p2]
        p -= 1
        p2 -= 1
```

```python
#Array Intersection → LC 349 – Intersection of Two Arrays
"""   interesection means- return the common elements from both the array
Input: nums1 = [1,2,2,1], nums2 = [2,2]
Output: [2] 

"""
## 1st common approach (complexity O(m+n))
def intersectionAr1(a1, a2):
    s1=set(a1)
    s2=set(a2)
    return list(s1.intersection(s2))
a1=[1,2,2,1]
a2=[2,2]
print(intersectionAr1(a1,a2))

##second approach  Using hash map()

def intersectionHashMap(a1,a2):
    seen=set(a1)
    intersect=set()
    for i in a2:
        if i in seen:
            intersect.add(i)
    return list(intersect)
print(intersectionHashMap(a1=a1,a2=a2))

## Two pointer approach  ( if both arrays are sorted)

def intersectionTwoPointer(a1,a2):
    result=set()
    i=j=0
    while i < len(a1) and j < len(a2):
        if a1[i]==a2[j]:
            result.add(a1[i]) ## check both are equal or not 
            i+=1
            j+=1
        elif a1[i] < a2[j]:
            i+=1 
        else:
            j+=1 
    return list(result)
a1=[1,2,2,1]
a2=[2,2]
print(intersectionTwoPointer(a1, a2))
    
```

```python
#Array Intersection → LC 349 – Intersection of Two Arrays
"""  Each element in the result must appear as many times as it shows in both arrays and you may return the result in any order.

Example 1:

Input: nums1 = [1,2,2,1], nums2 = [2,2]
Output: [2,2] , here 2 is present twice in both array """ 

# Best approach Hashmap time complexity ( O(m+n)) 
from collections import Counter
def intersectionTwiceHashmap(a1, a2):

    frquency=Counter(a1) ## count the frequency of each element
    intersct=[]

    for i in a2:
        if frquency[i]>0: ## if frequency of element found
            intersct.append(i)
            frquency[i]-=1
    return intersct

a1=[1,2,2,1]
a2=[2,2]
print(intersectionTwiceHashmap(a1, a2))
```

```python
## 32.	Balanced Parentheses → LC 20 – Valid Parentheses
"""Given a string s containing just the characters '(', ')', '{', '}', '[' and ']', determine if the input string is valid.

An input string is valid if:

Open brackets must be closed by the same type of brackets.
Open brackets must be closed in the correct order.
Every close bracket has a corresponding open bracket of the same type."""

def validParanthesis(s):
    while "()" in s or "{}" in s or "[]" in s:
        s=s.replace("{}", "")
        s=s.replace("[]","")
        s=s.replace("()","")
    return s==""
s="{[()]}"
print(validParanthesis(s))

## approach 2nd via stack (time - O(n) and space- O(n)){

""" Note- for stack solutions- 
Valid Parentheses
HTML tag validation
Expression evaluation
Removing invalid parentheses
Monotonic stack problems

"""
def invalid(s):
    stack=[]
    mapping={
        "}":"{",
        ")":"(",
        "]":"["
    }
    for char in s:
        if char in mapping: ## comparing for second element
            if not stack or stack[-1]!=mapping[char]:
                return False
            stack.pop()
        else:
            stack.append(char)
    return len(stack)==0
s="{[()]}"
print(invalid(s))

## approach 3  , to avoid using mapping
def isValid(s):
    stack = []

    for c in s:
        if c == '(':
            stack.append(')')
        elif c == '{':
            stack.append('}')
        elif c == '[':
            stack.append(']')
        else:
            if not stack or stack.pop() != c:
                return False

    #return not stack 
    return len(stack)==0
```
